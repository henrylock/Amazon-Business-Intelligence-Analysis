4. Cash flow

```python
# check null for cash flow
df_cash_flow.isna().sum()
```

The other null value aside from 2020 column are not removed. This is because financial statement are interconnected, and removing the rows with missing values might affects the balance of the accounting sheets and break ratio and trends.

```python
print(df_income_stmt.duplicated().sum())
print(df_quarter_income.duplicated().sum())
print(df_balance_sheet.duplicated().sum())
print(df_cash_flow.duplicated().sum())
```

```python
# transposing the dataset so the year are listed in one column and each observations (revenue, net debt etc) is across the columns

transp_income_stmt = df_income_stmt.transpose()
transp_quarter_income = df_quarter_income.transpose()
transp_balance_sheet = df_balance_sheet.transpose()
transp_cash_flow = df_cash_flow.transpose()
```

```python
# checking the data
df_income_stmt.head(5)
```

```python
# define a function to promote first row as header and reset the index
def promote_header(df, index_name='Year'):
    df.columns = df.iloc[0]                     # Promote first row to column headers
    df = df.drop(df.index[0])                   # Drop that row from data
    df.index.name = index_name                  # Rename index
    return df.reset_index()


# apply function all of the data
transp_income_stmt = promote_header(transp_income_stmt)
transp_quarter_income = promote_header(transp_quarter_income)
transp_balance_sheet = promote_header(transp_balance_sheet)
transp_cash_flow = promote_header(transp_cash_flow)
```

```python
# Correct the data type
# convert year column to datetime

transp_income_stmt['Year'] = pd.to_datetime(transp_income_stmt['Year'])
transp_quarter_income['Year'] = pd.to_datetime(transp_quarter_income['Year'])
transp_balance_sheet['Year'] = pd.to_datetime(transp_balance_sheet['Year'])
transp_cash_flow['Year'] = pd.to_datetime(transp_cash_flow['Year'])
```

```python
# checking the data
print(transp_income_stmt.head(5))
print(transp_income_stmt.info())
```

This datasets provides a comprehensive overview of Walmart and Costco in the United States,which are the direct competitors of Amazon covering the period from 2015 to 2025. There are 3 individual datasets which will be merged into one. These dataset contains revenue, market share and market spend data for the competitors.

```python
# read all 3 csv
df_costco= pd.read_csv('/content/drive/MyDrive/Colab Notebooks/Dataset/Costco_US_Revenue_and_Marketing_2015_2024.csv')
df_walmart= pd.read_csv('/content/drive/MyDrive/Colab Notebooks/Dataset/Walmart_US_Revenue_and_Marketing_2016_2025.csv')
df_walmart_cotsco = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/Dataset/Walmart_Costco_US_Market_Share_2016_2024.csv')


# print the first 5 rows for each dataset
print(df_costco.head(5))
print(df_walmart.head(5))
print(df_walmart_cotsco.head(5))

# print info for all datasets
print(df_costco.info())
print(df_walmart.info())
print(df_walmart_cotsco.info())
```

The datasets is clean and no null and duplicated value are found. The next step is to merge all three dataset using outer join to preserve the data in each datasets (Cotsco have data for 2015 while Walmart has data for 2025)

```python
# merging the three dataset into one using outer join
market_competitors = pd.merge(df_costco, df_walmart, on='Year', how='outer')
market_competitors = pd.merge(market_competitors, df_walmart_cotsco, on='Year', how='outer')

# checking the dataset
market_competitors.head(5)
```

## Merging the datasets into unified dataset in excel3
All the cleaned dataset is combined into one unified dataset in excel before used in PowerBI. First a dictionary for each sheet name with their rrespective dataframe is defined. Then, the output file is set. pd ExcelWriter is used to combine all the file into one excel file.

```python
!pip install xlsxwriter
```

```python

# Dictionary: sheet name → dataframe
dataframes = {
    'FBA US Location': df_usa,
    'FBA Canada Location': df_canada,
    'FBA Inventory Tool': df_inventory_tool,
    'Amazon Stocks': df_stocks,
    'Income Statement': transp_income_stmt,
    'Quarterly Income Statement': transp_quarter_income,
    'Balance Sheet': transp_balance_sheet,
    'Cash Flow': transp_cash_flow,
    'Competitors': market_competitors
}


# Output file
output_excel = 'clean_dataset_AA.xlsx'

# Write to Excel with multiple sheets
with pd.ExcelWriter(output_excel, engine='xlsxwriter') as writer:
    for sheet_name, df in dataframes.items():
        df.to_excel(writer, sheet_name=sheet_name[:31], index=False)
```
